# 2. List Comprenhension

**List Comprehension**
Permite crear una lista de Python basado en una sola linea.

```python
[i for i in range(9)]
```

```python
[str(i) for i in range(9)]
```

```python
[str(i) for i in range(9) if i % 2 == 0]
```

# 3. Instrucciones

**Instrucciones: Pass, Continue, Break**

```python
# Pass: No hace nada
def get_cociente(x, y):
    try:
        resultado = x // y
    except ZeroDivisionError:
      pass
```

```python
get_cociente(5, 0)
```

```python
for numero in range(5):
    if numero == 2:
        pass

    print('El numero es ' + str(numero))

print('Se terminó el loop')
```

```python
# Continue
for numero in range(5):
    if numero == 2:
        continue

    print('El numero es ' + str(numero))

print('Se terminó el loop')
```

```python
# Break
for numero in range(5):
    if numero == 2:
        break

    print('El numero es ' + str(numero))

print('Se terminó el loop')
```

# 4. Fechas

**Trabajando con Fechas**

```python
# Trabajando con fechas
from datetime import datetime, date, timedelta
```

```python
# # Fecha actual
fecha = date.today()
print(fecha)
```

```python
# # Fecha con hora (Ahora)
ahora = datetime.now()
print(ahora)
```

```python
# # Peruvian Time Zone
import pytz

peru_tz = pytz.timezone("America/Lima")
print(datetime.now(peru_tz))
```

```python
print(pytz.all_timezones)
```

```python
# # Sumando fechas: timedelta
datetime.now() + timedelta(weeks=13, days=1, hours=1, seconds=40)
```

```python
# # Sumando fechas: relativedelta
from dateutil.relativedelta import relativedelta
datetime.now() + relativedelta(years=2, months=25, weeks=13, days=1, hours=1, seconds=40)
```

```python
# De texto a datetime: strptime (p: parse)
date_object = datetime.strptime('05/08/2022 14:20:43', '%d/%m/%Y %H:%M:%S')
date_object
```

```python
# # Obteniendo fechas
date_object.date()
```

```python
# # Obteniendo horas
date_object.time()
```

```python
print(date_object.year, date_object.month, date_object.day, date_object.hour)
```

```python
# De datetime a texto: strftime (f: format)
date_object.strftime('%A %d de %B del %Y')
```

```python
date_object.strftime('%d-%b-%Y')
```

```python
# Formatos de datetime: https://www.w3schools.com/python/python_datetime.asp
```

```python
def to_fecha(date_object):
  dict_meses = {1:"Enero", 2: "Febrero", 3:"Marzo", 4:"Abril", 5:"Mayo", 6:"Junio", 7:"Julio", 8:"Agosto"}
  mes = dict_meses[date_object.month]
  return f'{date_object.day} de {mes} del {date_object.year}'
```

```python
parsed_date = to_fecha(date_object)
parsed_date
```
