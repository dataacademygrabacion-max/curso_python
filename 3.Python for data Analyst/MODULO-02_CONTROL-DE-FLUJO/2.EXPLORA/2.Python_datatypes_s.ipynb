```python
# Usando tuplas para multiples asignaciones
persona = ('Angel', 32, True)
nombre, edad, profesor = persona
print(nombre, edad, profesor)
```

```python
# Las tuplas generalmente son utilizadas para dar varios resultados en una función
def get_movil_details(movil):
  ruedas = ''
  puertas = ''

  if movil == 'moto':
    ruedas = 2
    puertas = 0

  return ruedas, puertas
```

```python
data = get_movil_details('moto')
data
```

```python
ruedas, puertas = get_movil_details('moto')
print(ruedas)
print(puertas)
```

```python
# Operaciones matemáticas en un set de Python
# # Encontrando diferencias entre dos sets
A = {1, 2, 3, 4, 5}
B = {3, 4, 5, 6, 7}
```

```python
print(A - B)
print(B - A)
```

```python
# # Encontrando la diferencia simetrica entre 2 sets
print(A ^ B)
```

```python
# # Encontrando la interseccion entre 2 sets
print(A & B)
```

```python
# Subsets y supersets
A = {1, 2, 3}
B = {1, 2, 3, 4, 5}
C = {1, 2, 3, 10}
```

```python
print(f'A < B: {A < B}')
print(f'C < B: {C < B}')
print(f'B > A: {B > A}')
print(f'B > C: {B > C}')
print(f'A < A: {A < A}')
print(f'A <= A: {A <= A}')
print(f'A >= A: {A >= A}')
```

# Diccionarios en Python

```python
# Creando un diccionario en blanco
diccionario = {}
otro_dicc = dict()
print(type(diccionario))
print(type(otro_dicc))
```

```python
# Accediendo a un diccionario
diccionario = {'key': 'valor de la llave key'}
print(diccionario)
print(diccionario['key'])
```

```python
dict_persona = {'Nombre':'Jose',
                'Apellido':'Perez',
                'Edad':20}
```

```python
# Accediendo a los elementos de un diccionario
for k,v in dict_persona.items():
    print(k, '-',v)
```

```python
# Agregando datos a un diccioanrio
dict_persona['Genero'] = 'M'
print(dict_persona)
```

```python
# Actualizando un valor en un diccionario
dict_persona['Edad'] = 30
print(dict_persona)
```

```python
# Eliminando datos de un diccionario
del(dict_persona['Genero'])
print(dict_persona)
```

```python
# # Keys
for dato in dict_persona.keys():
    print(dato)
```

```python
# # Values
for dato in dict_persona.values():
    print(dato)
```
