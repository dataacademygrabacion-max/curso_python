# Reto 02 — SOLUCION

No liberar hasta cerrar el modulo.

```python
ventas = [
    {"id": "V01", "producto": "Laptop",    "cantidad": 2, "precio": 3200.0, "ciudad": "Lima"},
    {"id": "V02", "producto": "Mouse",     "cantidad": 5, "precio": 65.0,   "ciudad": "Cusco"},
    {"id": "V03", "producto": "Laptop",    "cantidad": 1, "precio": 3200.0, "ciudad": "Lima"},
    {"id": "V04", "producto": "Monitor",   "cantidad": 3, "precio": 780.0,  "ciudad": "Arequipa"},
    {"id": "V05", "producto": "Teclado",   "cantidad": 4, "precio": 120.0,  "ciudad": "Lima"},
    {"id": "V06", "producto": "Mouse",     "cantidad": 8, "precio": 65.0,   "ciudad": "Arequipa"},
    {"id": "V07", "producto": "Monitor",   "cantidad": 2, "precio": 780.0,  "ciudad": "Cusco"},
    {"id": "V08", "producto": "Audifonos", "cantidad": 6, "precio": 210.0,  "ciudad": "Lima"},
    {"id": "V09", "producto": "Laptop",    "cantidad": 1, "precio": 3200.0, "ciudad": "Trujillo"},
    {"id": "V10", "producto": "Teclado",   "cantidad": 10, "precio": 120.0, "ciudad": "Cusco"},
]
```

## 1 — Importe

El diccionario es mutable: dentro del `for`, `v` apunta al mismo objeto de la lista,
asi que asignarle una clave modifica la lista original.

```python
for v in ventas:
    v["importe"] = v["cantidad"] * v["precio"]

assert ventas[0]["importe"] == 6400.0
print("OK ejercicio 1")
```

## 2 — Total por ciudad

`.get(ciudad, 0)` devuelve 0 la primera vez que aparece la ciudad, lo que evita
el `if ciudad not in por_ciudad`.

```python
por_ciudad = {}
for v in ventas:
    por_ciudad[v["ciudad"]] = por_ciudad.get(v["ciudad"], 0) + v["importe"]

assert round(por_ciudad["Lima"], 2) == 11340.0
print("OK ejercicio 2:", por_ciudad)
```

## 3 — Clasificar

`if / elif / else` en cascada: el orden importa. Si preguntaras primero por `>= 500`,
las ventas grandes caerian en MEDIANA.

```python
clasificadas = []
for v in ventas:
    if v["importe"] >= 2000:
        categoria = "GRANDE"
    elif v["importe"] >= 500:
        categoria = "MEDIANA"
    else:
        categoria = "CHICA"
    clasificadas.append(f"{v['id']} - {v['producto']} - S/ {v['importe']:.2f} - {categoria}")

assert clasificadas[0] == "V01 - Laptop - S/ 6400.00 - GRANDE"
assert clasificadas[1] == "V02 - Mouse - S/ 325.00 - CHICA"
for linea in clasificadas:
    print(linea)
```

## 4 — Producto top

Dos pasos: primero acumular, despues buscar el maximo. Mezclarlos en un solo bucle
tambien funciona, pero se lee peor.

```python
por_producto = {}
for v in ventas:
    por_producto[v["producto"]] = por_producto.get(v["producto"], 0) + v["importe"]

producto_top, monto_top = None, 0
for producto, monto in por_producto.items():
    if monto > monto_top:
        producto_top, monto_top = producto, monto

assert producto_top == "Laptop" and round(monto_top, 2) == 12800.0
print(f"Producto top: {producto_top} con S/ {monto_top:,.2f}")
print("\nAcumulado por producto:")
for p, m in sorted(por_producto.items(), key=lambda kv: -kv[1]):
    print(f"  {p:10} S/ {m:>10,.2f}")
```

## 5 — `while` con `break`

`while` cuando la condicion de parada NO es "recorri todo", sino un evento.
Aqui: parar al segundo GRANDE.

```python
i, impresas = 0, 0
while i < len(clasificadas) and impresas < 2:
    if clasificadas[i].endswith("GRANDE"):
        print(clasificadas[i])
        impresas += 1
    i += 1
```

### Errores tipicos al corregir

- Recalcular `cantidad * precio` en cada ejercicio en vez de reusar `v["importe"]`.
- Usar `sum()` sobre la lista completa en el ej. 2 y perder el corte por ciudad.
- Poner el `elif >= 500` antes que el `if >= 2000`.
- Formatear con `round(x, 2)` en vez de `f"{x:.2f}"`: `round(6400.0, 2)` imprime `6400.0`, no `6400.00`.
- Inicializar `monto_top = None` y romper la comparacion `>`.
