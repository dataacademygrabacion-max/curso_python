# 1. Booleanos y Condicionales

```python
Valores: True (V) o False (F)  ---> 1 v 0
Preposicion Logicas:

&& (Condicional Y) (and):
--> Para que sea True, todos los valores deben ser True, caso contrario es False
- V && V && V && V...--> V
- V && F --> F
- F && V --> F
- F && F --> F

|| (Condicional O) (or): Para que sea False, todos los valores deben ser False,
caso contrario es True
- V || V --> V
- V || F --> V
- F || V --> V
- F || F || F || F ...--> F

! (Negacion) (not): Cambia el valor de True a False y viceversa
- !V --> F
- !F --> V

edad = 18

Condicion: Validar si es mayor de edad: edad >= 18

edad >= 18
18   >= 18
True

!(edad >= 18) --> !True --> False

edad = 11

edad >= 18
11   >= 18
False

!(edad >= 18) --> !False --> True

Condicion: Validar si el valor de la edad es un numero correcto: edad < 0

edad = -20

edad < 0
-20 < 0
True
!(edad < 0) --> !True --> False

Operadores logicos:

>=/<= : Mayor igual que / Menor igual que
>/<   : Mayor que / Menor que
==    : Igual que
!=    : Diferente de

dni = "75286666"

Condicion: Validar que tenga 8 caracteres: (len(dni) == 8)

len(dni) == 8
8 == 8
True
!(len(dni) == 8) --> !True --> False

dni = "75286666"
len(dni) == 8
9 == 8
False
!(len(dni) == 8) --> !Fakse --> True
```

**El bucle WHILE:** "Mientras esta condición sea Verdadera, sigue ejecutando el código". ¡Cuidado con hacer bucles infinitos!

```python
condTrue = True
while condTrue:
   print("Mientras la condicion sea Verdadera")
```

```python
tope = 5
numero = 1
while numero <= tope:
    print(f'Este es el intento numero {numero}')
    numero = numero + 1

# Es vital sumar 1 al contador, de lo contrario será 1 por siempre.
```

# Lista []
Coleccion de datos organizadas (elementos), que puede crecer, encogerse o modificarse.


.append(elemento): Metodo exclusivo de las listas, que permite agregar un elemento al final de la lista

.remove(valor): Meotdo que busca un elemento especifico en la lista y lo elimina permanentemente.


len(coleccion): Te permite saber la cantidad de elementos.

max(coleccion): Funcion exclusiva que analiza una coleccion numerica, y devuelve el mayor valor

sum(coleccion): Funcion exclusica de una coleccion numerica que devuelva la suma.




```python
edades = [25, 30, 42, 18, 55]
edades.append(23)

total_clientes = len(edades)
print(f"Total de clientes: {total_clientes}")
```

```python
edades.remove(18)
edades
```

```python
carrito = [15.50, 40.00, 120.99, 800.00]
total_pagar = sum(carrito)
menor_precio = min(carrito)
max_precio = max(carrito)
print(f"""Total a pagar: ${total_pagar}\nMenor precio es: ${menor_precio}\nMayor precio es: ${max_precio}""")
```

# Diccionario {}
Una estructura organizada mediante pares clave: valor (clave tu encabezado : valor son los valores de tu columna)

.keys(): Extraer y devolver una coleccion (lista) de todas las columnas (clave)

diccionario[llave]: Permitir leer o modificar el valor de una etiqueta



```python
cliente_vip = {
    "id_cliente": 1001,
    "nombre": "Elena García",
    "membresia": "Premium"
}

nombre_extraido = cliente_vip["nombre"]
tipo_membresia = cliente_vip["membresia"]

print(f"El nombre del cliente es {nombre_extraido} y tiene la membresia {tipo_membresia}")
```
