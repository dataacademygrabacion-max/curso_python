El equipo de Marketing extrajo una lista de teléfonos de clientes desde una
página web, pero los usuarios escriben de formas muy raras
 (con guiones, espacios extra, etc.).
 Necesitamos una función que reciba esa lista sucia y devuelva una lista
 perfectamente limpia para subirla al sistema de mensajería (SMS).

```python
def limpiar_telefono(lista_telefonos):
  lista_numeros_limpios = []

  for numero in lista_telefonos:
    numero_limpio = numero.replace("-","").replace(" ","").strip()
    lista_numeros_limpios.append(numero_limpio)

  #lista_numeros_limpios = ["987654321","999888777","987987987"]
  return lista_numeros_limpios
```

```python
telefonos = ["  987-654-321","999 888 777","987-987-987  ","  987-654----321","999 888 777","987-987-987  "]

lista_numeros_limpios = limpiar_telefono(telefonos)
print(lista_numeros_limpios)
```

Recibimos la lista de ventas del día de un asesor comercial. La regla de la empresa es:

Si la venta es mayor a $1000, gana el 10% de comisión.

Si la venta está entre $500 y $1000, gana el 5%.

Si la venta es menor a $500, no gana comisión (0%).


Queremos una función que calcule cuánto dinero extra se llevará el asesor hoy.

```python
ventas_asesor_A = [1200, 300]
```

```python
def calcule_pago_comisiones(lista_ventas):
  total_comision = 0 #Inicializar el acumulador de comisiones

  for venta in lista_ventas:
    if venta > 1000:
      total_comision = total_comision + venta*0.1
    elif venta >= 500:
      total_comision = total_comision + venta*0.05

  return total_comision
```

```python
comision_asesor_A = calcule_pago_comisiones(ventas_asesor_A)
print(comision_asesor_A)
```

```python
def comision_venta(lista_venta):
  total_comision = 0
  for venta in lista_venta:
    if venta > 1000:
        total_comision = total_comision + venta * 0.1
    elif venta > 500 and venta < 1000:
        total_comision = total_comision + venta * 0.05

  return total_comision

lista_venta = [600, 1500.5, 501, 360]
comision = comision_venta(lista_venta)
print(f"El asesor ganará ${round(comision,2)} de comisión.")
```

# Manejo de Errores (try y except)

```python
numeros_denominador = [2,5,0,10]
numerador = 10
# fraccion--> numerador / denominador

for num in numeros_denominador:
  # Intentes hacer esto...
  try:
    resultado = numerador / num
    print(f'{numerador} / {num} = {resultado}')
  except:
    print(f'No se puede dividir {numerador} entre {num}')
```

```python
numeros = [2,5,"Gratis",10]
numerador = 10
# fraccion--> numerador / denominador

for num in numeros:
  # Intentes hacer esto...
  try:
    resultado = numerador / num
    print(f'{numerador} / {num} = {resultado}')
  except Exception as e:
    print(f'Hubo un error al procesar {num}. El sistema dice que el error fue: {e}')
```

# List Comprehension (Listas en una sola linea) -- FOR

[QUE_QUIERO_HACER   for  VARIABLE in LISTA_ORIGINAL]

```python
# El for clasico
numeros_viejos = [10,20,30]
numeros_nuevos = []  # Voy a sumarle +2 a cada elemento de lista vieja

for numero in numeros_viejos:
  numeros_nuevos.append(numero + 2)

print(numeros_nuevos)
```

```python
# El for en una linea
numeros_nuevos = [numero + 2 for numero in numeros_viejos]
print(numeros_nuevos)
```

```python
## Volver mayusculas los nombres de la lista

nombres_list = ["Ana","Erika","Yami"]
nombres_list_mayus = [ nombre.upper()  for nombre in nombres_list]
print(nombres_list_mayus)
```

# List Comprehension (Listas en una sola linea) -- IF

[QUE_QUIERO_HACER   for  VARIABLE in LISTA_ORIGINAL  if CONDICION]

```python
edades_list = [15,22,30,12]

mayor_edad = [ edad for edad in edades_list if edad >= 18]
print(mayor_edad)

menor_edad = [ edad for edad in edades_list if edad < 18]
print(menor_edad)
```
