# EXPLORA 07 — Tu primer repo, paso a paso

Practica guiada. El profesor va por el `_Caso`, tu vas por el `_Base`.

Al terminar tienes un repo local con historial, ramas y `.gitignore` — todo en una carpeta
temporal que puedes borrar sin miedo.

> Si `!git` no funciona en tu Jupyter, abre una terminal (Git Bash o Anaconda Prompt),
> navega a la carpeta con `cd` y escribe los mismos comandos sin el `!`.

```python
import tempfile, os
from pathlib import Path

TALLER = Path(tempfile.mkdtemp()) / "mi-primer-repo"
TALLER.mkdir(parents=True)
os.chdir(TALLER)
print(TALLER)
```

## Paso 1 — Inicializar

`-b main` fija el nombre de la rama principal. Sin eso, versiones viejas de git la
llaman `master` y no coincide con lo que espera GitHub.

```python
!git init -b main
# Config local del repo demo: asi la practica funciona aunque todavia no hayas
# hecho el 'git config --global'. En tus repos de verdad usa el global.
!git config user.name "Alumno Demo"
!git config user.email "demo@ejemplo.com"
!git status
```

## Paso 2 — Primer archivo y primer commit

```python
Path("README.md").write_text("# Analisis de ventas 2025\n", encoding="utf-8")
!git status --short
```

```python
!git add README.md
!git commit -m "Agrega README con el titulo del proyecto"
```

## Paso 3 — Historial

```python
!git log --oneline
```

## Paso 4 — Diff

El diff se mira **antes** de commitear. Es tu ultima oportunidad de ver que estas
guardando de verdad — y de detectar la contrasena que se colo sin querer.

```python
Path("README.md").write_text(
    "# Analisis de ventas 2025\n\n"
    "Consolida las ventas mensuales y responde 3 preguntas de negocio.\n",
    encoding="utf-8")
!git diff
```

```python
!git add README.md
!git commit -m "Describe el objetivo del analisis en el README"
```

## Paso 5 — `.gitignore`

```python
Path(".gitignore").write_text("*.csv\n.env\n.ipynb_checkpoints/\n", encoding="utf-8")
Path("datos.csv").write_text("a,b\n1,2\n", encoding="utf-8")
Path(".env").write_text("TOKEN=abc123\n", encoding="utf-8")

!git status --short
```

Solo aparece `.gitignore`. Los otros dos existen en disco pero git los ignora.

```python
!git add .gitignore
!git commit -m "Ignora datos y credenciales"
```

## Paso 6 — Rama

```python
!git switch -c nuevo-grafico
Path("grafico.py").write_text("import matplotlib.pyplot as plt\n", encoding="utf-8")
!git add grafico.py
!git commit -m "Agrega el script del grafico mensual"
!git log --oneline
```

```python
!git switch main
print("Archivos en main:", sorted(p.name for p in Path(".").iterdir() if p.name != ".git"))
```

`grafico.py` no esta. No se perdio: vive en la otra rama.

```python
!git merge nuevo-grafico
print("Archivos en main tras el merge:", sorted(p.name for p in Path(".").iterdir() if p.name != ".git"))
```

## Paso 7 — Conflicto

```python
!git switch -c titulo-corto
Path("README.md").write_text("# Ventas 2025\n", encoding="utf-8")
!git add -A && git commit -q -m "Acorta el titulo del README"

!git switch main
Path("README.md").write_text("# Analisis completo de ventas 2025\n", encoding="utf-8")
!git add -A && git commit -q -m "Amplia el titulo del README"

!git merge titulo-corto
```

```python
print(Path("README.md").read_text(encoding="utf-8"))
```

Ahi estan las marcas. Se edita a mano, se deja una version, se borran las marcas.

```python
Path("README.md").write_text("# Analisis de ventas 2025\n", encoding="utf-8")
!git add README.md
!git commit -m "Resuelve conflicto de titulo en el README"
```

## Paso 8 — Historial completo

```python
!git log --oneline --graph --all
```

---
### Lo que hay que llevarse

1. `git status` antes y despues de cada paso. Es gratis y te dice donde estas.
2. El diff se mira **antes** del commit, no despues.
3. `.gitignore` **antes** del primer `git add .`, no despues del susto.
4. Un conflicto no es un error: es git preguntandote cual version gana.
