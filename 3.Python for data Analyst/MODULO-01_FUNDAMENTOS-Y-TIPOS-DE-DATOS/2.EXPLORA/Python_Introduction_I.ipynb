**Manejo de Excepciones**

```python
# Excepciones comunes: NameError
pritn('test')
```

```python
try:
  pritn('test')
except:
  print('Esto es un error de NameError')
```

```python
# Excepciones comunes: KeyError
dict_alumno = {'nombre': 'Angel'}
dict_alumno['apellido']
```

```python
try:
  dict_alumno['apellido']
except:
  print('Esto es un error de KeyError')
```

```python
# Excepciones comunes: IndexError
lista = [1, 2, 3]
lista[10]
```

```python
try:
  lista[10]
except:
  print('Este es un IndexError')
```

```python
# Mejores practicas: No dejar el exception en blanco
```

```python
try:
  pritn('test')
except NameError:
  print('Esto es un error')


try:
  dict_alumno['apellido']
except KeyError as e:
  print('No se cuenta con la llave', e)

try:
  lista[10]
except IndexError as e:
  print('Este es un IndexError. Descripción:', e)
```

```python
# Exceptions: Completo, multiples except
def get_cociente(x, y):
    try:
        resultado = x // y
    except ZeroDivisionError:
      print("Error, estas dividiendo entre 0")
    except TypeError as e:
      print('Error de tipo TypeError. Detalle:', e)
    else:
      print("Tu respuesta es:", resultado)
    finally:
      print('Esto siempre se ejecuta')
```

```python
get_cociente(16, 5)
```

```python
get_cociente(16, 'a')
```

```python
get_cociente(16, 0)
```

```python
# Exceptions: Completo, multiples errores en un Except
def get_cociente(x, y):
    try:
        resultado = x // y
    except (ZeroDivisionError, TypeError) as e:
      print("E:", e)
    else:
      print("Tu respuesta es:", resultado)
    finally:
      print('Esto siempre se ejecuta')
```

```python
get_cociente(16, 5)
```

```python
get_cociente(16, 'a')
```

```python
get_cociente(16, 0)
```
