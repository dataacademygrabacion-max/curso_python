```python
#En el colab
url_archivo_colab_lg = "/content/Data/logistica_palote.txt"

#Leer el archivo
df_lg = pd.read_csv(
    url_archivo_colab_lg,
    sep="|"
    )

#Mostrar el archivo
display(df_lg)
```

```python
df_lg.dtypes
```

```python
df_lg["peso_kg"] = pd.to_numeric(df_lg["peso_kg"], errors="coerce")
```

```python
display(df_lg)
```

```python
# Imputacion de datos
# ¿Que hacemos con los nulos? Vamos a rellenarlo con el promedio de la serie
promedio_peso = df_lg["peso_kg"].mean()
print(promedio_peso)
df_lg["peso_kg"] = df_lg["peso_kg"].fillna(promedio_peso)
```

```python
display(df_lg)
```

```python
# ¿Como reemplazamos un dato en una serie? Utilizamos el replace
#Se imputa como entregado debido a que el area comercial indica que es por ...
data_imputar = "Entregado"

df_lg['estado'] = df_lg['estado'].replace("N/D",data_imputar)
display(df_lg)
```

```python
df_lg['estado'] = df_lg['estado'].str.lower()
display(df_lg)
```
