Setup

```python
import pandas as pd

data = {
    "fecha_venta": ["2024-01-05", "2024-01-10", None, "2024-02-01"],
    "producto": ["A101", "B202", "C303", "A101"],
    "cantidad": ["10", None, "5", "3"],  # cantidades como texto y con valores nulos
    "precio_unitario": ["20.5", "15.0", "30.0", "20.5"],  # precios como texto
    "sistema_id": [123, 456, 789, 999]  # columna irrelevante
}

df_raw = pd.DataFrame(data)
df_raw.to_csv("ventas_retailmax_raw.csv", index=False)

print("==> Dataset de ejemplo creado: ventas_retailmax_raw.csv")
df_raw.head()
```

Desarrollo

```python
import pandas as pd

df = pd.read_csv("ventas_retailmax_raw.csv")
```

```python
# 1. Convertir tipos de datos
df["fecha_venta"] = df["fecha_venta"].astype("datetime64[ns]")
df["cantidad"] = df["cantidad"].astype("Int64")
df["precio_unitario"] = df["precio_unitario"].astype("float")
```

```python
# 2. Manejar valores faltantes
df["cantidad"] = df["cantidad"].fillna(0)
df = df.dropna(subset=["fecha_venta"])
```

```python
# 3. Omitir columnas irrelevantes
df = df.drop(columns=["sistema_id"])
```

```python
print("==> Dataset limpio:")
print(df)
```
