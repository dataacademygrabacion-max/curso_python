```python
import pandas as pd
import os
import  shutil
```

```python
folders = ['server_inputs', 'server_outputs']
for folder in folders :
    if os.path.exists(folder):
        shutil.rmtree(folder)
    else :
        os.makedirs(folder, exist_ok=True)
```

# EXTRACT

```python
df=pd.read_csv('/content/drive/MyDrive/CURSOS/Diplomado DE/PYTHON DE/Proyecto final del curso/file.ope')
```

# TRANSFORM

```python
df['value']=df['Field_1'].astype('str').str[0]
```

```python
df_cliente=df[df.value=='1'].reset_index(drop=True).copy()
df_deuda=df[df.value=='2'].reset_index(drop=True).copy()
```

```python
df_sbs_cliente=df_cliente.Field_1.astype(str).str.split('|',expand=True).copy()
```

```python
col_cliente = {
    0: 'SBSCodigoCliente',
    1: 'SBSFechaReporte',
    2: 'SBSTipoDocumentoT',
    3: 'SBSRucCliente',
    4: 'SBSTipoDocumento',
    5: 'SBSNumeroDocumento',
    6: 'SBSTipoPer',
    7: 'SBSTipoEmpresa',
    8: 'SBSNumeroEntidad',
    9: 'SBSSalNor',
    10: 'SBSSalCPP',
    11: 'SBSSalDEF',
    12: 'SBSSalDUD',
    13: 'SBSSalAPER',
    14: 'SBSAPEPAT',
    15: 'SBSAPEMAT',
    16: 'SBSAPECAS',
    17: 'SBSNOMCLI',
    18: 'SBSNOMCLI2'
}
```

```python
df_sbs_cliente=df_sbs_cliente.rename(columns=col_cliente)
```

```python
df_deuda.head(5)
```

```python
df_deuda['CodigoSBS']=df_deuda['Field_1'].astype(str).str[0:10]
df_deuda['CodigoEmpresa']=df_deuda['Field_1'].astype(str).str[10:15]
df_deuda['TipoCredito']=df_deuda['Field_1'].astype(str).str[15:17]
df_deuda['Nivel2']=df_deuda['Field_1'].astype(str).str[17:19]
df_deuda['Moneda']=df_deuda['Field_1'].astype(str).str[19]
df_deuda['SubCodigoCuenta']=df_deuda['Field_1'].astype(str).str[20:31]
df_deuda['Condicion']=df_deuda['Field_1'].astype(str).str[31:37]
df_deuda['ValorSaldo']=df_deuda['Field_1'].astype(str).str[37:41]
df_deuda['ClasificacionDeuda']=df_deuda['Field_1'].astype(str).str[42]
df_deuda['CodigoCuenta']=df_deuda['Nivel2']+df_deuda['Moneda']+df_deuda['SubCodigoCuenta']


```

```python
# columans a nenombra
columns_deuda={'CodigoSBS':'Cod_SBS',
               'CodigoEmpresa':'Cod_Emp',
               'TipoCredito':'Tip_Credit',
               'ValorSaldo':'Val_Saldo',
               'CodigoCuenta':'Cod_Cuenta'}
```

```python
df_deuda=df_deuda.rename(columns=columns_deuda)
```

```python
df_deuda.columns
```

```python
df_sbs_deuda=df_deuda.copy()
```

```python
del df_sbs_deuda['Field_1'] ,df_sbs_deuda['value']
```
