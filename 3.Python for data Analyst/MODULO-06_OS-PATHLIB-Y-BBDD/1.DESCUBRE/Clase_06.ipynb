#Enumerate
Sirve para recorrer elementos de una lista y tener su indice (la posicion) al mismo tiempo, no es necesario crear variables de contadores manuales.
Por defecto el indice inicia en 0

Permite que sea mas limpio y rapido el codigo.

```python
import pandas as pd
import numpy as np
```

```python
datos = {
    'sucursal': ['Lima', 'Lima', 'Cusco', 'Cusco', 'Arequipa', 'Lima', np.nan],
    'vendedor': ['Ana', 'Luis', 'Ana', 'Carlos', 'Luis', 'Ana', 'Pedro'],
    'producto': ['Laptop', 'Mouse', 'Laptop', 'Teclado', 'Mouse', 'Monitor', 'Teclado'],
    'ventas_usd': [1200, 50, 1200, 80, 50, 300, 80],
    'cantidad': [1, 2, 1, 3, 2, 1, 3]
}

df = pd.DataFrame(datos)
display(df)
```

```python
sucursales = ['Lima', 'Cusco', 'Arequipa']
for sucursal in sucursales:
    print(sucursal)
```

```python
for sucursal in enumerate(sucursales):
    print(sucursal)
```

```python
for indice, sucursal in enumerate(sucursales):
    print(f'Sucursal #{indice+1}: {sucursal}')
```

```python
for ciudad in enumerate(sucursales, start=2):
  print(ciudad)
```

```python
for posicion, ciudad in enumerate(sucursales, start=1):
  print(f'Sucursal #{posicion}: {sucursal}')
```

#groupby (Excel: Tabla dinamica pero sin ser dinamica)
SPLIT: Dividir la tabla originales en pequeños grupos basados en los valores de una o mas columnas.

APPLY: Aplicamos una funcion de agregacion, es una funcion matematica que se le aplica a cada grupo de manera independiente.

COMBINE: Junta todos los resutlados en una tabla resumida.

Te va permitir responder preguntas de negocio.

Metodos:

*   .sum(): suma
*   .mean(): promedio
*   .count(): recuento
*   .agg(): te permite utilizar mas de una funcion de agregacion

```python
datos = {
    'sucursal': ['Lima', 'Lima', 'Cusco', 'Cusco', 'Arequipa', 'Lima', np.nan],
    'vendedor': ['Ana', 'Luis', 'Ana', 'Carlos', 'Luis', 'Ana', 'Pedro'],
    'producto': ['Laptop', 'Mouse', 'Laptop', 'Teclado', 'Mouse', 'Monitor', 'Teclado'],
    'ventas_usd': [1200, 50, 1200, 80, 50, 300, 80],
    'cantidad': [1, 2, 1, 3, 2, 1, 3]
}

df = pd.DataFrame(datos)
display(df)
```

```python
ventas_por_sucursal = df.groupby('sucursal').sum()

print(ventas_por_sucursal)
```

```python
ventas_por_sucursal_isnum = df.groupby('sucursal').sum(numeric_only=True)

print(ventas_por_sucursal_isnum)
```

```python
df.groupby(['sucursal', 'vendedor']).agg({
                                        'ventas_usd': 'sum',
                                        'cantidad': ['count','sum']
                                        })
```

```python
ventas_por_sucursal = df.groupby('sucursal')['ventas_usd'].sum()

print(ventas_por_sucursal)
```

```python
ventas_por_sucursal_agg = df.groupby('sucursal').agg({'ventas_usd':'sum'})

print(ventas_por_sucursal_agg)
```

```python
report_conNa = df.groupby(
    ['sucursal','vendedor','producto'],
    dropna=False
).agg({'ventas_usd':'sum'})

print(report_conNa)
```

```python
report_conNa = df.groupby(
    ['sucursal','vendedor','producto'],
    dropna=False,
    as_index=False
).agg({'ventas_usd':'sum'})

print(report_conNa)
```

#Enumerate con groupby

```python
list_gb_ciudades = df.groupby('sucursal')

for ciudad in enumerate(list_gb_ciudades):
  print(ciudad)
```

```python
for indice,(nombre_ciudad,sub_df) in enumerate(list_gb_ciudades):
  print(f'\nCiudad #{indice+1}: {nombre_ciudad}')
  print(sub_df[['vendedor','ventas_usd']])
```

# np.where
Es una funcion de la libreria Numpy que se comporta como un condicional para un arreglo o lista.

```python
import pandas as pd
import numpy as np
```

```python
datos = {
    'estudiantes':['Ana','Luis','Carlos','Maria','Pedro','Sofia'],
    'nota': [18,10,15,8,np.nan,20],
    'asistencias': [90,85,40,50,10,100],
    'edad':[22,17,25,16,21,20]
}
df = pd.DataFrame(datos)
```

```python
display(df)
```

```python
df['estado'] = np.where(
    df['nota']>= 11,  # Condicion
    'Aprobado',       # En caso sea verdadero la condicion
    'Reprobado'       # En caso sea falsa la condicion
)
```

```python
display(df)
```

```python
df['beca'] = np.where(
    (df['edad']>= 11) & (df['nota']>=18),  # Condicion mas de una or --> |
    'Otorgado',       # En caso sea verdadero la condicion
    'Denegado'       # En caso sea falsa la condicion
)
```

```python
display(df)
```

```python
df['rendimiento'] = np.where(df['nota']>17, 'Sobresaliente',
         np.where(df['nota']>=11, 'Regular', 'Deficiente')
         )
```

```python
display(df)
```

```python
df['nota_final'] = np.where(
    df['asistencias'] < 50,
    df['nota']*0.8,
    df['nota']
)
```

```python
display(df)
```
