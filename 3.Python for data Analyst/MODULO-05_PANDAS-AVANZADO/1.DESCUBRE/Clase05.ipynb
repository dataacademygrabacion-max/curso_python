# Limpieza - Eliminación de duplicados

```python
df_empleados = pd.DataFrame({
    "salario_txt": ["S/. 4500", "USD 1200", "Sin Dato", "S/. 3200"]})
```

```python
def limpiar_salario(salario_txt):
    if "S/. " in salario_txt:
        salario = salario_txt.replace("S/. ","")
        return float(salario)
    elif "USD" in salario_txt:
        salario = salario_txt.replace("USD ","")
        return float(salario) * 3.5
    else:
        return 0.0
```

```python
df_empleados["salario_float"] = df_empleados["salario_txt"].apply(limpiar_salario)
```

```python
df_empleados
```

```python
df_ventas_regionales = pd.DataFrame({
    "ciudad": ["Lima", "Cusco", "Lima", "Arequipa"],
    "monto": [150.0, 200.0, 50.0, 300.0]
})
```

```python
def calcular_comision(fila):
  ciudad = fila["ciudad"]
  monto = fila["monto"]

  if ciudad == "Lima":
    return monto * 0.1
  elif ciudad != "Lima":
    return monto * 0.05
```

```python
df_ventas_regionales["comision"] = df_ventas_regionales.apply(calcular_comision, axis=1)
```

```python
df_ventas_regionales
```

```python
# Simulamos dos tablas
df_clientes = pd.DataFrame({
    "id_cliente": [1, 2, 3],
    "nombre": ["Ana", "Luis", "Carlos"]
})

df_ventas = pd.DataFrame({
    "id_cliente": [1, 2, 4], # Nota: No hay venta del cliente 3, y el 4 es un cliente fantasma
    "monto": [500, 300, 900]
})

# 1. INNER JOIN (La Intersección estricta)
# Solo trae los que hacen 'Match' perfecto en ambas tablas (Clientes 1 y 2)
df_inner = pd.merge(df_clientes, df_ventas, on="id_cliente", how="inner")

# 2. LEFT JOIN (Manda la Izquierda)
# Trae TODOS los clientes. Si no tienen venta, Pandas les pondrá NaN (Cliente 3)
df_left = pd.merge(df_clientes, df_ventas, on="id_cliente", how="left")

# 3. RIGHT JOIN (Manda la Derecha)
# Trae TODAS las ventas. Si la venta no tiene cliente registrado, le pone NaN (Cliente 4)
df_right = pd.merge(df_clientes, df_ventas, on="id_cliente", how="right")

# 4. OUTER JOIN (El Todo Incluido)
# Trae a TODOS, crucen o no crucen. (Clientes 1, 2, 3 y 4)
df_outer = pd.merge(df_clientes, df_ventas, on="id_cliente", how="outer")

print("Ejemplo de INNER JOIN (La Intersección estricta):")
display(df_inner)
print("Ejemplo de LEFT JOIN (Manda la Izquierda):")
display(df_left)
print("Ejemplo de RIGTH JOIN (Manda la Derecha):")
display(df_right)
print("Ejemplo de OUTER JOIN (El Todo Incluido):")
display(df_outer)
```
