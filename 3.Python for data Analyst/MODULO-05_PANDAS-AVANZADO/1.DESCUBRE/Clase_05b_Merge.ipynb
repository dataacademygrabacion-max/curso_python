```python
import pandas as pd

datos_restaurantes = {
    'id_del_restaurante': [1,2],
    'name': ['La parrilla', 'Sushi'],
    'category': ['Carnes', 'Comida Japonesa']
}
df_restaurantes = pd.DataFrame(datos_restaurantes)
```

```python
datos_menus = {
    'id_menu': [101,102,103,104],
    'restaurante_id': [1,1,2,3],
    'name': ['Menu 1', 'Menu 2', 'Menu 3', 'Menu 4'],
    'category': ['Almuerzo', 'Almuerzo', 'Cena', 'Cena']
}
df_menus = pd.DataFrame(datos_menus)
```

```python
df_master_problema = pd.merge(df_restaurantes, df_menus, on='id_restaurante')
print(df_master_problema)
```

```python
df_master_sufijos = pd.merge(
    df_restaurantes,
    df_menus,
    on='id_restaurante',
    suffixes=('_restaurantes', '_menus')
    )
display(df_master_sufijos)
```

```python
df_master_sufijos_2 = pd.merge(
    df_menus,
    df_restaurantes,
    left_on='restaurante_id',
    right_on='id_del_restaurante',
    suffixes=('_rest', '_platos'),
    how='left'
    )
display(df_master_sufijos_2)
```
