```python
headers = {'x-api-key': 'reqres-free-v1'}
```

```python
post_url = "https://reqres.in/api/users"
data = {
    "name": "morpheus",
    "job": "leader"
}
response = requests.post(post_url, headers=headers, json=data)
print(response.status_code)
response.json()
```

```python
# PUT 
#BORRA TODA Y SOLO DEJA EL ID
post_url = "https://reqres.in/api/users/2" # -> ESPECIFICA QUE DATA MODIFICAR
data = {
    "name": "JHON",
    "job": "zion resident"
}
response = requests.put(post_url, headers=headers, json=data)
print(response.status_code)
response.json()
```

```python
# PATCH
post_url = "https://reqres.in/api/users/2"
data = {
    "name": "morpheus",
    "job": "zion resident"
}
response = requests.patch(post_url, headers=headers, json=data)
print(response.status_code)
response.json()
```
