Ej 1. Realizar el equivalente en Python del siguiente query:
===
```sql
select
FUENTE,
min(CODIGO) as MIN_CODIGO,
max(CODIGO) as MAX_CODIGO,
count(DPTO) as COUNT_DPTO
from df_base
group by
FUENTE
order by
MIN_CODIGO ASC,
MAX_CODIGO ASC,
COUNT_DPTO DESC

```

```sql
select
FUENTE,
DPTO,
```
