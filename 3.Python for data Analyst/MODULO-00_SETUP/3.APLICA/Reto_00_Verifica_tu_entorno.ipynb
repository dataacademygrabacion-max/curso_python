## 3. Pandas lee un archivo

Los datos del curso viven en `_RECURSOS/datasets/`. Ajusta la ruta relativa segun
desde donde abriste el notebook.

**Ojo:** `ventas.csv` usa `;` como separador y `,` como decimal. Si no lo indicas,
pandas te devuelve una sola columna.

```python
ruta = "../../_RECURSOS/datasets/ventas.csv"

ventas = pd.read_csv(ruta, sep=";", decimal=",")
print("Filas y columnas:", ventas.shape)
ventas.head()
```

## 4. Matplotlib dibuja

```python
fig, ax = plt.subplots(figsize=(6, 3))
ventas["ciudad"].value_counts().plot(kind="bar", ax=ax)
ax.set_title("Ventas por ciudad")
ax.set_ylabel("N de ventas")
plt.tight_layout()
plt.show()
```
