**Matploit**


1.   Grafico de lineas (plot): Mostrar tendencias a lo largo del tiempo.
2.   Grafico de dispersion (scatter): Mostrar una relacion o correlacion visual entre dos variables numericas. Identifacr patrones.
3.   Grafico de barras (bar): Comparar cantidades  entre diferentes categorias.
4.   Histograma (hist): Mostrar la distribucion de una variable numerica para identificar donde se concentran los datos.
5. Grafico circular o de pastel (pie): Mostrar proporciones de un todo (100%)
6. Grafico de caja (boxplot): sirve para mostrar un resumen estadistico (mediana, cuartiles, promedio) y detectar valores atipicos (outliers)



```python
import matplotlib.pyplot as plt
import numpy as np
```

```python
x = [1,2,3,4,5]
y = [2,3,7,1,0]
```

```python
plt.plot(x,y)
plt.show()
```

```python
plt.plot(x,y,
         color = 'green',
         linestyle = '--', # solida (-) , discontinua (--), punteada (:)
         linewidth = 2, # grosor de la linea
         marker = '^', # Punto (o), cuadrado (s), triangulo (^)
         markersize = 10, # tamaño del marcador
         label = 'Ventas - Proyeccion'
         )
plt.title('Ventas Mensuales (Enero - Mayo)')
plt.xlabel('Meses')
plt.ylabel('Ventas (MM PEN)')
plt.grid(True)
plt.legend()
plt.show()
```

```python
y2 = [3,2,8,3,1] #Ventas reales

plt.plot(x,y, label = 'Ventas - Proyeccion', marker = '^',linestyle = '--', color = 'green')
plt.plot(x,y2, label = 'Ventas - Real', marker = 's', linestyle = '-', color = 'black',)
plt.annotate('Deviacion - NC',
                  xy = (2,2), # señalar el punto en la grafica
                  xytext = (2.5,1), # posicion del texto
                  arrowprops = dict(
                      facecolor = 'green',
                      arrowstyle = '->')
)
plt.title('Ventas Mensuales (Enero - Mayo)')
plt.xlabel('Meses')
plt.ylabel('Ventas (MM PEN)')
plt.grid(True)
plt.legend()
plt.show()
```

```python
x = np.random.rand(50)
y = np.random.rand(50)

plt.scatter(x,y)
plt.show()
```

```python
plt.scatter(x,y,
            s=100,
            c= 'purple',
            alpha = 0.5,
            marker = '*')
plt.title('Dispercion con Transparencia')
plt.show()
```

```python
colores = np.random.rand(50)
tamanios = 1*np.random.rand(50) * 100

scatter = plt.scatter(x,y,
            s=tamanios,
            c=colores,
            alpha = 0.5,
            marker = '*',
            cmap = 'inferno') #

plt.colorbar(scatter, label = 'Intensidad')
plt.show()
```

```python
categorias  = ['A','B','C','D']
valores = [25,40,15,30]

plt.title('Grafico de barras verticales')

plt.bar(categorias, valores)
plt.show()
```

```python
plt.barh(categorias, valores, color = 'green', edgecolor = 'navy', height = 0.6)
plt.title('Grafico de barras horizontal')
plt.show()
```

```python
valores2 = [10,20,10,15]
plt.bar(categorias, valores, label = 'Serie 1', color = 'b')
plt.bar(categorias, valores2,bottom = valores, label = 'Serie 2', color = 'r')

plt.title('Grafico de barras apiladas')
plt.xlabel('Categorias')
plt.ylabel('Valores')
plt.legend()
plt.show()
```

```python
datos = np.random.randn(1000)

plt.hist(datos)
plt.show()
```

```python
plt.hist(datos, bins = 50, color = 'orange',edgecolor = 'black',alpha = 0.7)
plt.title('Histograma de distribucion de datos')
plt.xlabel('Rango de valores')
plt.ylabel('Frecuencia')
plt.show()
```

```python
plt.hist(datos, bins = 10, color = 'orange',edgecolor = 'black',alpha = 0.7)
plt.title('Histograma de distribucion de datos')
plt.xlabel('Rango de valores')
plt.ylabel('Frecuencia')
plt.show()
```

```python
datos2 = np.random.randn(1000) #Distribucion desplazada

plt.hist(datos, bins = 30, color = 'orange',edgecolor = 'black',alpha = 0.7, label = 'Dist 1', density = True)
plt.hist(datos2, bins = 30, color = 'blue',edgecolor = 'black',alpha = 0.7, label = 'Dist 2', density = True)

plt.title('Histograma de distribucion de datos')
plt.xlabel('Rango de valores')
plt.ylabel('Frecuencia')
plt.legend()
plt.show()
```

```python
proporciones = [30,20,40,10]
etiquetas = ['Manzanas','Peras','Naranjas','Uvas']

plt.pie(proporciones, labels = etiquetas)
plt.show()
```

```python
separacion = (0.1,0,0,0) # solo separa la primera rebanada
plt.pie(proporciones, labels = etiquetas,
        explode = separacion, # 0 = unido, >0 = separado
        autopct='%1.1f%%', # formato del porcentaje: que muestre un decimal
        shadow = True, # sombra 3D
        startangle=90, # angulo de inicio, es para girar el grafico
        )
plt.show()
```

```python
plt.pie(proporciones, labels =etiquetas, autopct ='%1.1f%%', pctdistance=0.75)

#Graficamos un ciculo blanco en el centro para hacer la dona
centro = plt.Circle((0,0), 0.5, color = 'white')
fig = plt.gcf() # figura actual
fig.gca().add_artist(centro) # añadir el circulo a los ejes actuales
plt.show()
```

```python
datos_caja = np.random.randn(100) # 100 numeros aleatorios

plt.boxplot(datos_caja)
plt.show()
```

```python
grupo1 = np.random.normal(100,10,200)
grupo2 = np.random.normal(90,20,200)

plt.boxplot([grupo1,grupo2],labels=['Grupo A','Grupo B'],
            vert=False, # hace el grafico horizontal
            patch_artist=True #permite rellenar la caja de color
            )
plt.show()
```

```python
box_props = dict(facecolor = 'lightblue', color = 'blue',linewidth = 2)
median_props = dict(color = 'red', linewidth = 2)
whisker_props = dict(color = 'green',linestyle = '--' ,linewidth = 1.5)
flier_props = dict(marker = 'o',markerfacecolor = 'yellow', markersize = 8, markeredgecolor = 'black')

plt.boxplot([grupo1,grupo2],labels=['Grupo A','Grupo B'],patch_artist=True,
            boxprops = box_props,
            medianprops = median_props,
            whiskerprops = whisker_props,
            flierprops = flier_props
            )
plt.title("Boxplot Personalizado")
plt.show()
```
